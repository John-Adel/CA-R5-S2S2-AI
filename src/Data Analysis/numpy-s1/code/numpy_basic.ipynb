```python
class Student:

    def __init__(self, name):
        self._name = name
    @property #getter cause name is protected
    def name(self):
        return self._name
    @name.setter
    def name(self, name):
        self._name = name
```

```python
def array_reshape(arr, rows: int = -1, cols: int = -1): # we used defaults as -1 because -1 gets the last number of a dimention of a matrix which returns the same shape
    try:
        arr = np.array(arr)
        return arr.reshape(rows, cols)
    except (TypeError, ValueError) as e:
        raise ValueError(f"Invalid reshape operation: {e}")
    
```

```python
vec = [1, 2, 3, 4, 5, 6]
array_reshape(vec, 2, 3)
```

```python
# Higher priority overrides the lower one
boolean = np.array([[False, False, 1],
                [True, True, True],
                [True, False, False]])
print(boolean.dtype)
print(boolean, "\n")

mat = np.array([[1, 2, 3], # here data type is int
                [4, 5, 6],
                [7, 8, 9]])
print(mat.dtype)
print(mat, "\n")

mat1 = np.array([[1.5, 2, 3], # this is float which has a higher data type priority than int
                [4, 5, 6],
                [7, 8, 9]])
print(mat1.dtype)
print(mat1, "\n")

mat2 = np.array([[1.4, "2", 3], # this is string which has a higher data type priority than float
                [4, 5, 6],
                [7, 8, 9]])
print(mat2.dtype)
print(mat2, "\n")

mat3 = np.array([[1.4, "2", o1],
                [4, 5, 6],
                [7, 8, 9]])
print(mat3.dtype)
print(mat3, "\n")
```
